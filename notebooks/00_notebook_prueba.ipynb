## 1. Versión de Python

## 2. Librerías instaladas

Importamos las librerías base y mostramos sus versiones. Si alguna falla al
importar, revisá que hayas corrido `pip install -r requirements.txt` con el
entorno virtual activado.

```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns

print("pandas:    ", pd.__version__)
print("numpy:     ", np.__version__)
print("matplotlib:", matplotlib.__version__)
print("seaborn:   ", sns.__version__)
```

## 3. Un DataFrame de prueba

Creamos una tabla pequeña con pandas para confirmar que podemos manipular datos.

```python
df = pd.DataFrame({
    "mes": ["Ene", "Feb", "Mar", "Abr", "May"],
    "ventas": [120, 135, 100, 160, 180],
})
df
```

## 4. Un gráfico de prueba

Graficamos esos datos para confirmar que la visualización funciona.

```python
plt.figure(figsize=(6, 4))
plt.plot(df["mes"], df["ventas"], marker="o")
plt.title("Ventas por mes (datos de prueba)")
plt.xlabel("Mes")
plt.ylabel("Ventas")
plt.grid(True, alpha=0.3)
plt.show()
```
